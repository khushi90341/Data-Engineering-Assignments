#  3.Handle missing values (identify, fill/drop).

In this step, missing values are identified and handled using appropriate methods.

```python
# Checking missing values in dataset
df.isnull().sum()
```

```python
# Filling numeric missing values with mean
df["rating"] = df["rating"].fillna(df["rating"].mean())
```

```python
df["ratings_count"] = df["ratings_count"].fillna(0)
```

```python
# Text column example
df["seller_name"] = df["seller_name"].fillna("Unknown")
```

```python
df["what_customers_said"] = df["what_customers_said"].fillna("No review")
```

```python

df.drop(["videos", "variations", "seller_information"], axis=1, inplace=True)
```

```python
# fill missing values in discount column with 0 
df["discount"] = df["discount"].fillna(0)
```

```python
# Checking missing values again after handling
df.isnull().sum()
```

# 4.Remove duplicates

In this step, duplicate rows are identified and removed from the dataset.

```python
# Checking duplicate rows
df.duplicated().sum()
```

```python
# Removing duplicate rows
df = df.drop_duplicates()
```

#  5.Perform basic operations (filter rows, select columns). 


In this step, basic operations such as selecting columns and filtering rows are performed on the dataset.

```python
# Selecting first two columns
df[['title', 'final_price']]
```

```python

# Filtering rows using numeric column
high_rated = df[df['rating'] > 4]
high_rated.head()
```

# Correct Final Price
applying discount on initial_price which was not appliead in given csv

```python
df['final_price'] = (
    df['initial_price']
    - (df['initial_price'] * df['discount'] / 100)
)
```

```python
df.head()
```

# 6.Create a derived column (total_amount = price * quantity).

In this step, a new column named total_amount is created using price and quantity columns.

```python
# Creating temporary price and quantity columns

df['price'] = df['final_price']
```

```python
df['quantity'] = 1
```

```python
# Creating derived column
df['total_amount'] = (
    df['final_price'] * df['quantity']
)
```

```python
#Check Derived Column
df[['title', 'final_price', 'quantity', 'total_amount']].head()
```

#  7: Save the Cleaned Dataset as a New CSV File

In this step, the cleaned dataset is saved as a new CSV file for future analysis.

```python
## Saving cleaned dataset
df.to_csv(
    "cleaned_ecommerce_dataset.csv",
    index=False
)
```
